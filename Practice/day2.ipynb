### Q1. Second Largest Number

**Rule: Don't use sort(), max(), or set().**

````python
numbers = list(map(int,input("Enter math marks of everyone \"Note\" : use \',\' during speration.\n").split(',')))
print(f"You enter: {numbers}")

for j in range(len(numbers)):
    for i in range(len(numbers)-1):
        if numbers[i] < numbers[i+1]:
            numbers[i],numbers[i+1] = numbers[i+1], numbers[i]

print(f"Math\'s\' marks in decending order is : {numbers}") 
for i in range(len(numbers)):
    if numbers[i] > numbers[i+1]:
        secondLarger = numbers[i+1]
        break
 
print(f"Second largest number is : {secondLarger}")        
````

### Q4. Check Palindrome

````python
name = input("Enter your name : ")

name_reverse = ""

for i in range(len(name)-1,-1,-1):
    name_reverse = name_reverse + name[i]


print(f"Origional name : {name}")
print(f"reverse name : {name_reverse}")

palindrome = True
for i in range(len(name)):
    if name[i].lower() != name_reverse[i].lower():
       
        palindrome = False
        break
if palindrome:
    print(f"Your name is palindrome")
else:
    print("Your name is not plaindrome")    
````

### Q5. Count Vowels

1. Given a list of geographic feature names (e.g., river names, city names, village names), count how many names start with a vowel.
2. A GIS database contains names of water bodies. Calculate the total number of vowels present across all feature names for linguistic analysis.

````python
# Task 1 of question 5

name = list(map(str, input("Enter your city,village and nearest river name, \"note : Use \',\' for speration of names\"").split(',')))
print(f"You enter : {name}")
vowel_list = ['a','i','e','o','u']
vowel_count = 0
for i in name:
    if i[0].lower() in vowel_list:
        vowel_count += 1
print(f"{vowel_count} names starts with a vowel")      


# Task 2
river_name = list(map(str, input("Enter your state river names, \"note : Use \',\' for speration of names\"").split(',')))
river_vowel_count = 0
print(f"You enter these rivers:  {river_name}")

for river in river_name:
    for i in river:
        if i.lower() in vowel_list:
            river_vowel_count += 1
print(f"{river_vowel_count} vowel occurs in your all river characters.")            
````
