**Explanation:**

* 3 appears 2 times

* 7 appears 1 time ← first non-repeating

* 8 appears 2 times

### Electric Pole Inspection Report

```
poles = [
    {"Pole_ID": "P101", "Status": "Good"},
    {"Pole_ID": "P102", "Status": "Damaged"},
    {"Pole_ID": "P103", "Status": "Good"},
    {"Pole_ID": "P104", "Status": "Replace"},
    {"Pole_ID": "P105", "Status": "Damaged"},
    {"Pole_ID": "P106", "Status": "Good"},
    {"Pole_ID": "P107", "Status": "Replace"}
]
```

**Find :**
1. Total Poles
2. Good
3. Damaged
4. Replace
5. Percentage Good
6. Percentage Damaged
7. Percentage Replace
8. Action Required

**Rule**
Damaged + Replace >= 3 then Action Required: YES

**Insights :**
1. How many poles need maintenance?
2. What percentage of assets are damaged?
3. Do we need a field repair crew?
