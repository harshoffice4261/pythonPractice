### Q1. Second Largest Number

**Rule: Don't use sort(), max(), or set().**

````python
numbers = list(map(int,input("Enter math marks of everyone \"Note\" : use \',\' during speration.\n").split(',')))
print(f"You enter: {numbers}")

for j in range(len(numbers)):
    for i in range(len(numbers)-1):
        if numbers[i] < numbers[i+1]:
            numbers[i],numbers[i+1] = numbers[i+1], numbers[i]

print(f"Math\'s\' marks in decending order is : {numbers}") 
for i in range(len(numbers)):
    if numbers[i] > numbers[i+1]:
        secondLarger = numbers[i+1]
        break
 
print(f"Second largest number is : {secondLarger}")        
````

### Q2. Reverse a List

**Rule: Don't use:**
1. reverse()
2. [::-1]

````python
row = list(map(int,input("Enter roll no of row 5 \"Note\" : use \',\' during speration.\n").split(',')))

print(f"Row 5 have these roll no. from N-TO-S : {row}")



row_Reverse = []
for i in range(len(row)-1,-1,-1):
    
    row_Reverse.append(row[i])

print(f"Row 5 have these roll no. from S-TO-N : {row_Reverse}")    

````

### Q3. Find Common Elements by using loops.

````python
north_feeder_consumers = list(map(int,input("Enter consumer no. \"Note\" : use \',\' during speration.\n").split(',')))

south_feeder_consumers = list(map(int,input("Enter consumer no. \"Note\" : use \',\' during speration.\n").split(',')))

print(f"North Consumer is : {north_feeder_consumers}")
print(f"South Consumer is : {south_feeder_consumers}")


# method 1

# north_feeder_consumers = set(north_feeder_consumers)
# south_feeder_consumers = set(south_feeder_consumers)

# print(f"Common consumer : {north_feeder_consumers.intersection(south_feeder_consumers)}")

# method 2

common_consumer = []
for i in north_feeder_consumers:
    for j in south_feeder_consumers:
        if i==j and i not in common_consumer:
            common_consumer.append(i)

print(f"Common consumer : {common_consumer}")          
````
