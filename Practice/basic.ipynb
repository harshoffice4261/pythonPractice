### Q1 Find the Largest,Lowest and Average Number


```python
alititude = list(map(float,input("Enter the highest points where you had climbed : ").split(',')))

high = alititude[0]
low = alititude[0]
avg = 0

for i in alititude:
    if high < i:
        high = i
    if low > i:
        low = i  

    avg = avg + i    
        

print(f"Your highest climbed allititude is {high}") 
print(f"Your lowest climbed allititude is {low}") 
print(f"Your average climbed allititude is {avg/len(alititude)}") 
```

###  Q2. Count Vowels in "programming"

```python
# word = "progrAmming"
word = input("Enter any word: ")
word = word.lower()

print(f"\'A\' : {word.count('a')}")
print(f"\'E\'/ : {word.count('e')}")
print(f"\'I\' : {word.count('i')}")
print(f"\'O\' : {word.count('o')}")
print(f"\'U\' : {word.count('u')}")

```

### Q4. Even and Odd Counter

```python
numbers = list(map(int,input("Enter some numbers (seperated by \',\'): ").split(',')))

for i in numbers:
    if i>0 :
        if i%2 == 0:
            print(f"{i} : Even")
        else:
            print(f"{i} : Odd")  
    else :
        print(f"{i} : Not an integer")         
```
