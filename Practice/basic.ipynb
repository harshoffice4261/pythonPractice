### Q1 Find the Largest,Lowest and Average Number


```python
alititude = list(map(float,input("Enter the highest points where you had climbed : ").split(',')))

high = alititude[0]
low = alititude[0]
avg = 0

for i in alititude:
    if high < i:
        high = i
    if low > i:
        low = i  

    avg = avg + i    
        

print(f"Your highest climbed allititude is {high}") 
print(f"Your lowest climbed allititude is {low}") 
print(f"Your average climbed allititude is {avg/len(alititude)}") 
```

### Q4. Even and Odd Counter

```python
numbers = list(map(int,input("Enter some numbers (seperated by \',\'): ").split(',')))

for i in numbers:
    if i>0 :
        if i%2 == 0:
            print(f"{i} : Even")
        else:
            print(f"{i} : Odd")  
    else :
        print(f"{i} : Not an integer")         
```

### Q8. FizzBuzz (Classic Interview Question)

Print numbers from 1 to 20.

**Rules:**

Multiple of 3 → Fizz

Multiple of 5 → Buzz

Multiple of both → FizzBuzz

```python
for i in range(1,21,1):
    if i%3==0 and i%5==0:
        print(f"{i} : FizzBuzz")
    elif i%3==0:
        print(f"{i} : Fizz")
    elif i%5==0:
        print(f"{i} : Buzz")       
```

### Q10. Mini ATM
Initial balance: 10,000

**Menu:**
1. Check Balance
2. Deposit
3. Withdraw
4. Exit

```python


ATM_Numbers = {
    1001: 1250,
    1002: 5000,
    1003: 320,
    1004: 8750,
    1005: 15000,
    1006: 980,
    1007: 4500,
    1008: 22000,
    1009: 7600,
    1010: 100
}

request = True
account = 0
while request:
    account = int(input("Plz enter your account number : "))
    if account in ATM_Numbers.keys():
        print(f"Welcome {account}")
        break
    else:
        print("Your details is not fount! Plz try again.")
        request = True

userQuery = int(input(" Enter 1 for \"Check Balance\"\n Enter 2 for \"Deposit\"\n Enter 3 for \"Withdraw\"\n Enter 4 for \"Exit\"\n"))

if userQuery == 1:
    print(f"Your balance is : {ATM_Numbers[account]} rupees")
elif userQuery == 2:
    depositAmount = int(input("Enter your amount that you want to deposit : "))
    print(f"Amount before deposit : {ATM_Numbers[account]}")
    ATM_Numbers[account] += depositAmount
    print(f"Amount after deposit : {ATM_Numbers[account]}")
elif userQuery == 3:
    withdrawAmount = int(input("Enter your amount that you want to withdraw : "))
    balanceNew = ATM_Numbers[account] - withdrawAmount
    if balanceNew>=0:
        print(f"Amount before withdraw : {ATM_Numbers[account]}")
        ATM_Numbers[account] -= withdrawAmount
        print(f"Amount after withdraw : {ATM_Numbers[account]}")   
    else:
        print("You did not have sufficient balance!")   
elif userQuery ==4:
    print(f"{account} Thanks for coming")        
else: 
    print("You enter wrong input")



```

### Given:
**numbers = [12, 5, 12, 8, 20, 5, 1]**

**Try to find:**
1. Largest number
2. Smallest number
3. Sum of all numbers
4. Average
5. Remove duplicates
6. Sort ascending
7. Sort descending
8. Count how many times each number appears

```python
numbers = [12,5,12,8,20,5,1]

# Task 1,2,3 & 4
largest = numbers[0]
smallest = numbers[0]
sum = 0
for i in numbers:
    if i>largest:
        largest=i
    if smallest>i:
        smallest=i
    sum += i   
print(f"Largest number is : {largest}")   
print(f"Smallest number is : {smallest}")     
print(f"Sum of all elements is : {sum}")    
print(f"Average of all elements is : {sum/len(numbers)}") 


   


```
